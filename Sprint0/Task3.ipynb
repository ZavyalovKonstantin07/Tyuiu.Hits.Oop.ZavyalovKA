<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Invoice в C#, который будет представлять информацию о
фактурах за поставленные товары или оказанные услуги. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.

Требования к базовому классу Invoice:

• Атрибуты: Номер фактуры (InvoiceNumber), Дата выдачи (IssueDate), Общая
сумма (TotalAmount).

• Методы:

o CalculateTotal(): метод для расчета общей суммы по фактуре.

o AddLine(LineItem lineItem): метод для добавления позиции в фактуру.

o RemoveLine(LineItem lineItem): метод для удаления позиции из
фактуры.

Требования к производным классам:
1. ТоварнаяФактура (GoodsInvoice): Должна содержать дополнительные
атрибуты, такие как Дата поставки (SupplyDate). Метод AddLine() должен
быть переопределен для добавления информации о дате поставки товара
при добавлении позиции.
2. УслуговаяФактура (ServiceInvoice): Должна содержать дополнительные
атрибуты, такие как Дата оказания услуги (ServiceDate).
Метод RemoveLine() должен быть переопределен для добавления
информации о причине аннулирования услуги при удалении позиции.
3. КомбинированнаяФактура (CombinedInvoice) (если требуется третий класс):
Должна содержать дополнительные атрибуты, такие как Наличие возврата
(ReturnAllowed). Метод CalculateTotal() должен быть переопределен для
учета возможного возврата товара или услуги при расчете общей суммы.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System; 
 
class LineItem 
{ 
    public string Name; 
    public double Price; 
 
    public LineItem(string name, double price) 
    { 
        Name = name; 
        Price = price; 
    } 
} 
 
class Invoice 
{ 
    public int InvoiceNumber; 
    public DateTime IssueDate; 
    public double TotalAmount; 
 
    protected LineItem[] lines = new LineItem[10]; 
    protected int count = 0; 
 
    public virtual void CalculateTotal() 
    { 
        TotalAmount = 0; 
 
        for (int i = 0; i < count; i++) 
        { 
            if (lines[i] != null) 
            { 
                TotalAmount += lines[i].Price; 
            } 
        } 
    } 
 
    // virtual позволяет переопределить метод 
    // в дочернем классе 
    public virtual void AddLine(LineItem lineItem) 
    { 
        lines[count] = lineItem; 
        count++; 
    } 
 
    public virtual void RemoveLine(LineItem lineItem) 
    { 
        for (int i = 0; i < count; i++) 
        { 
            if (lines[i] == lineItem) 
            { 
                lines[i] = null; 
                break; 
            } 
        } 
    } 
} 
 
class GoodsInvoice : Invoice 
{ 
    public DateTime SupplyDate; 
 
    // Переопределяем AddLine() (полиморфизм) 
    public override void AddLine(LineItem lineItem) 
    { 
        lines[count] = lineItem; 
        count++; 
 
        Console.WriteLine("Товар добавлен."); 
        Console.WriteLine("Дата поставки: " + SupplyDate.ToShortDateString()); 
    } 
} 
 
class ServiceInvoice : Invoice 
{ 
    public DateTime ServiceDate; 
    public string CancelReason; 
 
    // Переопределяем RemoveLine() (полиморфизм) 
    public override void RemoveLine(LineItem lineItem) 
    { 
        for (int i = 0; i < count; i++) 
        { 
            if (lines[i] == lineItem) 
            { 
                lines[i] = null; 
                break; 
            } 
        } 
 
        Console.WriteLine("Услуга удалена."); 
        Console.WriteLine("Причина: " + CancelReason); 
    } 
} 
 
// Тест 
 
// Товарная фактура 
GoodsInvoice goods = new GoodsInvoice(); 
 
goods.InvoiceNumber = 1; 
goods.IssueDate = DateTime.Now; 
goods.SupplyDate = DateTime.Now; 
 
// Первый товар 
LineItem phone = new LineItem("Телефон", 50000); 
LineItem headphones = new LineItem("Наушники", 5000); 
 
goods.AddLine(phone); 
goods.AddLine(headphones); 
goods.CalculateTotal(); 
 
Console.WriteLine("Сумма товарной фактуры: " + goods.TotalAmount); 
 
 
// Услуговая фактура 
ServiceInvoice service = new ServiceInvoice(); 
 
service.InvoiceNumber = 2; 
service.IssueDate = DateTime.Now; 
service.ServiceDate = DateTime.Now; 
service.CancelReason = "Услуга больше не нужна"; 
 
// Первая услуга 
LineItem repair = new LineItem("Ремонт", 10000); 
LineItem cleaning = new LineItem("Чистка компьютера", 3000); 
 
service.AddLine(repair); 
service.AddLine(cleaning); 
service.CalculateTotal(); 
 
Console.WriteLine("Сумма услуговой фактуры: " + service.TotalAmount); 
 
// Удаляем услугу 
service.RemoveLine(repair);
```
